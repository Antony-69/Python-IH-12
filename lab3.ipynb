Написати програму, яка аналізує комплексне число my_number і
виводить в термінал одне з наступні значення:

o "уявна частина більша ніж дійсна частина на <різниця>", якщо
уявна частина більша ніж дійсна.

o "уявна частина менша ніж дійсна частина на <різниця>", якщо
уявна частина менша ніж дійсна.

o "уявна і дійсна частини рівні", якщо обидві рівні.

o "уявна і дійсна частини рівні нулю", якщо обидві рівні нулю.
Примітка: в трикутних дужках <> вивести відповідне значення.
Перевірити чи вірно виконуються всі комбінації.

```python
a = float(input("Enter the real part: "))
b = float(input("Enter the imaginary part: "))

my_number = complex(a, b)
print("Complex number:", my_number)

if my_number.imag > my_number.real:
    print(f"The imaginary part is greater than the real part by {my_number.imag - my_number.real}")
elif my_number.real > my_number.imag:
    print(f"The real part is greater than the imaginary part by {my_number.real - my_number.imag}")
elif my_number.imag == my_number.real:
    print("The imaginary part is equal to the real part.")
elif my_number.imag == 0 and my_number.real == 0:
    print("The complex number is zero.")
```

Задано список [2, 8, 0, -3, 8, -7, 4, 2, 5, -1, 8, 0, 5, -1, 2, 0, 1]
Написати програму, яка рахує кількість чисел менших ніж 0, більших ніж
0 та рівних 0 та виводить результати підрахунку, а також нові 3
посортовані списки отримані в результаті сортування.


```python
my_list = [2, 8, 0, -3, 8, -7, 4, 2, 5, -1, 8, 0, 5, -1, 2, 0, 1]
lower = 0
greater = 0
equal = 0

for n in my_list:
    if n < 0:
        lower += 1
    if n > 0:
        greater += 1
    if n == 0:
        equal += 1

print (lower, greater, equal)
```

Задано списки A = [1, 3, 5], B = [0, 5, 10, 20], та C = [1, 2]
Написати програму, яка формує список D, що містить добутки елементів
списку А, на елементи списку B для всіх можливих комбінацій та список
E який містить добутки елементів всіх трьох вхідних списків.
Для прикладу список D повинен бути: [0, 5, 10, 20, 0, 15, 30, 60, 0, 25, 50,
100].

```python
A = [1, 3, 5]
B = [0, 5, 10, 20]
C = [1, 2]

D = []
E = []

for a in A:
    for b in B:
        D.append(a * b)

for a in A:
    for b in B:
        for c in C:
            E.append(a * b * c)

print("D:", D)
print("E:", E)
```
