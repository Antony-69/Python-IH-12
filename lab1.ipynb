Написати умову при якій людина має право отримувати соціальну
допомогу за заданими чотирма вхідними параметрами: age, profit,
children_all_count, children_small_count. Тут age - вік особи, profit –
сімейний дохід, chіldren_all_count та children_small_count - кількість
всіх дітей та в тому числі неповнолітніх відповідно. Самому
придумати критерії за якими надається соціальна допомога.
Перевірити умову для різних комбінацій вхідних даних

```python
age = int(input("Enter your age: "))
profit = float(input("Enter your family income: "))
children_all_count = int(input("Enter the total number of children: "))
children_small_count = int(input("Enter the number of underaged children: "))

minimum_income = 3000
income_limit = 3000 + children_small_count * minimum_income

if children_small_count > children_all_count:
    print("The number of minor children cannot exceed the total number of children.")
else:
    eligible = (18 <= age <= 65 and profit < income_limit)

    print(f"Age: {age}; income: {profit}; income limit: {income_limit}")
    print("Social assistance is ELIGIBLE" if eligible else "You are a rich one mf, u don't need no social assistance")


limit = 3000 + children_small_count * 3000
valid_counts = children_small_count <= children_all_count
result = valid_counts and 18 <= age <= 65 and profit < limit
print(
    f"age={age}, profit={profit}, children={children_all_count}, "
    f"minor={children_small_count}: {'eligible' if result else 'not eligible'}"
)
```
