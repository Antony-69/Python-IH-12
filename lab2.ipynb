 Створити текстову змінну "6.28" та привести її до типів int та float.
Вивести в термінал результати

```python
a = "6.28"
print(f"float of \"{a}\" is {float(a)}")
print(f"int of \"{a}\" is {int(float(a))}")
```

Створити список А з довільних цілих чисел. Створити інший пустий
список В. Почергово додати в список В текстові стрічки, які отримані
приведенням відповідних елементів масиву А в текстові значення.

```python
A = [3, 5, 7, 9]
B = []

for element in A:
    B.append(str(element))

print("A:", A)
print("B:", B)
```

Створити список А з довільних різних чисел. Посортувати його в
порядку зростання. Отриманий посортований список доповнити його
(додати в кінець) інвертованим списком самого себе (Інвертований -
порядок елементів (reverse)). Наприклад, список А після сортування
рівний [1,4,8,9], а після доповнення [1,4,8,9,9,8,4,1].

```python
A = [5, 3, 7, 9]
print("A:", A)

A.sort()
print("A:", A)

A.extend(A[::-1])
print("A:", A)
```
