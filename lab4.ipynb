Задано кортеж ("середня", "школа", 75). Написати програму, яка в
циклі витягає елементи кортежу та об'єднує їх в текстовий рядок,
вставивши пробіли між словами та зробивши перетворення типів де
необхідно. Результат вивести в термінал. 

```python
tuple1 = ("middle", "school", 75)
print(tuple1)
str1 = ""
for i in tuple1:
    str1 += str(i) + " "
print(str1)
```

Задано множину {"apple", "banana", "cherry", "orange", "potatoes",
"plum"}. Написати програму, яка створює кортеж зі слів, які містяться
в цій множині і містять літеру "l" або “p” та ще один кортеж, зі слів,
які містяться в цій множині і містять літеру "l" та “p”. Очевидно, що
не вручну вибрати слова і з них зробити кортеж, а це повинна робити
програма. Для перевірки програми можете модифікувати вміст
кортежу та перевірити результат.


```python
set1 = {"apple", "banana", "cherry", "orange", "potatoes", "plum"}
tuple1 = ()
tuple2 = ()
for i in set1:
    if "l" in i or "p" in i:
        tuple1 += (i,)
    if "l" in i and "p" in i:
        tuple2 += (i,)
print (tuple1)
print (tuple2)
```

Задано два списки K та V однакової довжини. K = ["k1", "k2", "k3",
"k4"], V = ["10", 20, "40", 30]. Написати програму, яка створює
словник, в якому в якості ключів взяті елементи списку К, а відповідні
їх значення взяті зі списку V. В отриманому словнику проаналізувати
всі значення ключів і для ключів, значення, яких є типу str (стрічка),
здійснити їх заміну на тип int(цілі). Підказка: if (str == type(x))

```python
K = ["k1", "k2", "k3", "k4"]
V = ["10", 20, "40", 30]
dict1 = {}
for i in range(len(K)):
    dict1[K[i]] = V[i]
print (dict1)

for key, value in dict1.items():
    print(f"v {value} is {type(value)}")
    if type(value) == str:
        value = int(float(value))
    dict1[key] = value
print (dict1)
```
