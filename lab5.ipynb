Задано список A = [9, -1, 8, -2, 10, -5, 6, -8, 0, -3, 1]. Написати програму,
яка здійснює нормування елементів списку, до значення, яке введене
користувачем за допомогою команди input

```python
A = [9, -1, 8, -2, 10, -5, 6, -8, 0, -3, 1]

limit = float(input("Insert a value for normalization: "))

max_abs = max(abs(value) for value in A)

if max_abs == 0:
    normalized = [0 for value in A]
else:
    normalized = [value * limit / max_abs for value in A]

print("A:", A)
print("Normalized list:", normalized)
```

Задано список A = [9, 1, 8, 2, 10, 5, 6, 8, 0, 3, 1]. Написати програму, яка
вилучає зі списку один найбільший та один найменший елементи. Інші
залишити без змін їх взаємного розташування (не пересортовувати).
Результат помістити в інший список B та вивести в термінал.

```python
A = [9, 1, 8, 2, 10, 5, 6, 8, 0, 3, 1]

max_index = A.index(max(A))
min_index = A.index(min(A))

# Remove the first occurrence of the maximum and the first occurrence of the minimum.
B = [value for index, value in enumerate(A) if index not in (max_index, min_index)]

print("A:", A)
print("B:", B)
```

Написати програму, яка в термінал виводить таблицю ASCII великих та
малих літер латиниці у форматі типу як показано нижче:\
порядковий номер - символ - код (10-ва сис) - код (16-ва сис)\
00  A    065     0x41\
01  B    066     0x42\
02  C    067     0x43\
і т.д.\
Після літери "Z" перестрибнути на малу літеру "a", та продовжити до "z".\
Примітка: Додати передуючі нулі з допомогою функції format, щоб
порядковий номер містив 2 цифри, а код в десятковій системі - 3 цифри.

```python
number = 0

for code in range(ord('A'), ord('Z') + 1):
    print(format(number, '02d'), chr(code), format(code, '03d'), hex(code))
    number += 1

for code in range(ord('a'), ord('z') + 1):
    print(format(number, '02d'), chr(code), format(code, '03d'), hex(code))
    number += 1
```
